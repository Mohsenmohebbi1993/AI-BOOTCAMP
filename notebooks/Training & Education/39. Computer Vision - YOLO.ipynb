```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python

# chnage BGR to RGB

def show_bgr(img_bgr , title="Result"):
    """Change GBR to RGB
    """
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 6))
    plt.imshow(img_rgb)
    plt.title(title)
    plt.axis("off")
    plt.show()



```

```python
model = YOLO("yolov8n.pt")
results = model(PATH_IMG)
```

```python
PATH_IMG = r"C:\Mohsen Folder\Ai Bootcamp\Ai Bootcamp\data\raw\Training & Education Data\CV\test_3.jpg"

CONF = 0.2

results = model(PATH_IMG, conf=CONF)

img_annotated = results[0].plot()

show_bgr(img_annotated, title=f"YOLO detections (conf = {CONF})")
```
