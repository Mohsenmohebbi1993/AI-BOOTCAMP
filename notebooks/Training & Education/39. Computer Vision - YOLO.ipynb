```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python

# chnage BGR to RGB

def show_bgr(img_bgr , title="Result"):
    """Change GBR to RGB
    """
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(10, 6))
    plt.imshow(img_rgb)
    plt.title(title)
    plt.axis("off")
    plt.show()



```
