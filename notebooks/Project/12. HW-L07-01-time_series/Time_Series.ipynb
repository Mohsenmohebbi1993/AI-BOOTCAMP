```python
# step 1.1. Add data as dataframe
print("="*40)
print("step 1.1. Add data as dataframe")
print("="*40)

df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

# chnage type (date)
df_train["date"] = pd.to_datetime(df_train["date"])
df_test["date"] = pd.to_datetime(df_test["date"])

print("data:\n  train.csv\n  test.csv ")
```

Create one Function for print Information of dataFrame

```python
# Data Description
def data_description(dataframe, name="dataframe"):
    """ ## data_description:
    ###  print:   
    1. Description   
    2. Information   
    3. Shape   
    4. Name of Columns   
    5. Type each Column   
    6. Count Null value   
    7. Start, End, Total Days and Month and Years
    """
    print("="*40)
    print(f"Description {name}".center(40," "))
    print("="*40)
    print("")

    # info
    print(f"info from {name}".center(40,"-"))
    print(dataframe.info())
    print("")
    
    # shape
    print("="*40)
    print(f"shape from {name}".center(40,"-"))
    print(f"df_trane:\n   columns:{dataframe.shape[0]}\n    rows:{dataframe.shape[1]}")
    print("")

    # columns name
    print("="*40)
    print(f"columns name from {name}".center(40,"-"))
    print(dataframe.columns.to_list())
    print("")

    # Type each column
    print("="*40)
    print(f"Type of columns from {name}".center(40,"-"))
    print(dataframe.dtypes)
    print("")

    # sum of null
    print("="*40)
    print(f"null value of columns from {name}".center(40,"-"))
    print(dataframe.isna().sum())
    print("")

    # start and end date
    print("="*40)
    print(f"Start and End and Time Period Date from {name}".center(40,"-"))
    print(f"  Start date:         {dataframe["date"].min().strftime("%Y/%m/%d")}")
    print(f"  End date:           {dataframe["date"].max().strftime("%Y/%m/%d")}")
    print(f"  Total Date:         {(df_train["date"].max() - df_train["date"].min()).days + 1} days")
    print(f"  Total Months Date:  {df_train["date"].dt.to_period("M").nunique()} Months")
    print(f"  Total Years Date:   {df_train["date"].dt.to_period("Y").nunique()} Years")
    print("")


```

```python
data_description(df_train, "df_train")
```

**Check whether all dates between the first and last dates are present in the dataset.**

Miss Dayes are:
- 1 : 2013/12/25
- 2 : 2014/12/25
- 3 : 2015/12/25
- 4 : 2016/12/25

```python

print("="*40)
print("Are all days present in both the train and test ?")

# all daye from train and test
dayes = (pd.concat([df_train["date"], df_test["date"]])
         .drop_duplicates()
         .reset_index(drop=True)
         )

ALL_DAYE = pd.date_range(
    start=dayes.min(),
    end=dayes.max(),
    freq="D"
    )

miss_dayes = ALL_DAYE.difference(dayes)
if len(miss_dayes) != 0:
    print(f"Miss Dayes are:")
    for i in range(len(miss_dayes)):
        print(f">>> {i+1} : {miss_dayes[i].strftime("%Y/%m/%d")}")
else:
    print("There is NOT Miss dayes")
```
